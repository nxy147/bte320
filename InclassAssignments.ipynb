Recursive Function: String Reversal

```python
def reverse(s):
    if len(s) == 0:
        return ""
    return reverse(s[1:]) + s[0]

print(reverse("hello"))   # olleh
print(reverse("12345"))   # 54321
```

Part B - Monthly Loan Payment PMT

```python
def loan_payment(principal, annual_rate, years):
    assert (isinstance(principal, (int, float)) and
            isinstance(annual_rate, (int, float)) and
            isinstance(years, (int, float)))

    if (principal <= 0) or (annual_rate <= 0) or (years <= 0):
        raise ValueError("invalid input")

    try:
        r = annual_rate / 12
        n = years * 12
        if r == 0:
            pmt = principal / n
        else:
            pmt = principal * r / (1 - (1 + r) ** (-n))
    except Exception as e:
        print(f"Error: {e}")

    return round(pmt, 2)


print(loan_payment(200000, 0.06, 30))
print(loan_payment(10000, 0.05, 5))
```

Classes Point and Rectangle

* create a class point that has:
two coordiante attributes (x,y)
* A method distance() that calcs the distance between two instances of the class

* Create a class Rectangle that has:
Six attributes: four point instances nad two attributes for width//heieght. The latter two are calculated based onp airs of the four Point instances
* thes

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    def distance(self, other):
        dx = (self.x - other.x) ** 2
        dy = (self.y - other.y) ** 2
        return (dx + dy) ** 0.5

class Rectangle:
    def __init__(self, p1, p2, p3, p4):
        self.p1 = p1
        self.p2 = p2
        self.p3 = p3
        self.p4 = p4
        self.width = self.width_()
        self.height = self.height_()

    def width_(self):
        return self.p1.distance(self.p2)

    def height_(self):
        return self.p1.distance(self.p3)

    def area(self):
        return self.width * self.height

p1 = Point(0, 0)
p2 = Point(5, 0)
p3 = Point(0, 5)
p4 = Point(5, 5)

z = Rectangle(p1, p2, p3, p4)  # pass the four points in
z.area()
```

```python

class Employee:
    def __init__(self, name, wage, hours_worked):
        self.name = name
        self.__wage = None
        self.set_wage(wage)
        self.hours_worked = hours_worked

    def get_wage(self):
        return self.__wage

    def set_wage(self, wage):
        if wage < 0:
            print("Error: Wage cannot be negative. Wage unchanged.")
        else:
            self.__wage = wage

    def calculate_pay(self):
        return self.get_wage() * self.hours_worked

    def display_info(self):
        print(f"Name: {self.name}")
        print(f"Hourly Wage: ${self.get_wage():.2f}")
        print(f"Hours Worked: {self.hours_worked}")
        print(f"Weekly Pay: ${self.calculate_pay():.2f}")


class Manager(Employee):
    def __init__(self, name, wage, hours_worked, bonus):
        super().__init__(name, wage, hours_worked)
        self.bonus = bonus

    def calculate_pay(self):
        return super().calculate_pay() + self.bonus

    def display_info(self):
        super().display_info()
        print(f"Bonus: ${self.bonus:.2f}")
        print(f"Total Pay (with bonus): ${self.calculate_pay():.2f}")



print("===== Employee =====")
emp = Employee("Noah Yi", 18.50, 40)
emp.display_info()

print("\n===== Manager =====")
mgr = Manager("Charlie Baer", 35.00, 40, 500)
mgr.display_info()


print("\n--- Encapsulation Demo ---")
print(f"get_wage(): ${emp.get_wage():.2f}")
emp.set_wage(-5)
emp.set_wage(20)
print(f"Updated wage: ${emp.get_wage():.2f}")
```
