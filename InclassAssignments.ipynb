Design a simple numerical calc. in python. The user provides three inputs: two numbers and the mathematical operator of their choice. The program should print the output on the screen

```python
num1 = float(input("Enter the first number: "))
operator = input("Enter the operator (+, -, *, /): ")
num2 = float(input("Enter the second number: "))
if operator == "+":
  result = num1 + num2
elif operator == "-":
  result =  num1 - num2
elif operator == "*":
  result = num1 * num2
elif operator == "/":
  if(num2 == 0):
    result = ("Error: Division by zero is not allowed.")
  else:
    result = num1 / num2
print(result)
```
